# Adult dataset

```python
vanilla_results = pd.read_csv("../../../falsb4mpa/results/adult-mpa-bin-wout-agg-vanilla_lr-100.csv")
multi_adv_results = pd.read_csv("../../../falsb4mpa/results/adult-mpa-bin-wout-agg-mult_adv-100.csv")
```

## Gender

### Vanilla

```python
adult_df = vanilla_results
```

```python
female_metrics = adult_df[["a1_TN_g0","a1_FP_g0","a1_FN_g0","a1_TP_g0"]]
num_females = female_metrics.sum(axis=1)
male_metrics = adult_df[["a1_TN_g1","a1_FP_g1","a1_FN_g1","a1_TP_g1"]]
num_males = male_metrics.sum(axis=1)
```

```python
female_pos_rate = female_metrics["a1_FP_g0"] + female_metrics["a1_TP_g0"]
male_pos_rate = male_metrics["a1_FP_g1"] + male_metrics["a1_TP_g1"]
dif_pos_rate = abs(female_pos_rate/num_females - male_pos_rate/num_males)
```

```python
female_neg_rate = female_metrics["a1_FN_g0"] + female_metrics["a1_TN_g0"]
male_neg_rate = male_metrics["a1_FN_g1"] + male_metrics["a1_TN_g1"]
dif_neg_rate = abs(female_neg_rate/num_females - male_neg_rate/num_males)
```

```python
imparity = (dif_pos_rate + dif_neg_rate)/2
imparity
```

```python
imparity.groupby(imparity.index // 5).mean()
```

## Viz

```python
models = [
    "Vanilla",
    "Adversarial",
    "InfoFair",
    "Vanilla",
    "MultiAdvDP",
    "MultiAdvEqOdds",
    "MultiAdvEqOpp"
]

ours = [
    "[31]",
    "[31]",
    "[31]",
    "Ours",
    "Ours",
    "Ours",
    "Ours"
]
```

```python
imparity_gender_adult = [
    0.066,
    0,
    0.047,
    0.185508,
    0.209359,
    0.208561,
    0.216599,
]
```

```python
imparity_race_adult = [
    0.062,
    0,
    0.042,
    0.099533,
    0.167760,
    0.165384,
    0.166397
]
```

```python
micro_f1_adult = [
    0.83,
    0.743,
    0.816,
    0.8403,
    0.8335,
    0.8338,
    0.8311,
]
```

```python
df = pd.DataFrame({
    'models': models,
    'micro_f1': micro_f1_adult,
    'imparity_gender': imparity_gender_adult,
    'imparity_race': imparity_race_adult,
    'source': ours
})
df
```

```python
plt.style.use(["science", "ieee", "no-latex"])

sns.set_theme(
    style="whitegrid",
    context="paper"
)

palette = {
    '[31]': '#0072B2',
    'Ours': '#D55E00'
}

markers = {
    '[31]': '.',
    'Ours': '*'
}

fig, ax = plt.subplots(figsize=(7, 5))

sns.scatterplot(
    data=df,
    x='micro_f1',
    y='imparity_gender',
    hue='source',
    style='source',
    markers=markers,
    palette=palette,
    s=100,
    linewidth=0.8,
    edgecolor='white',
    ax=ax
)

for _, row in df.iterrows():
    if row["models"] == "MultiAdvDP":
        ax.annotate(
            row['models'],
            xy=(row['micro_f1'], row['imparity_gender']),
            xytext=(32, 8),
            textcoords='offset points',
            ha='left',
            va='center',
            fontsize=8.5,
            arrowprops=dict(
                arrowstyle='-',
                color='0.4',
                lw=0.7,
                shrinkA=5,
                shrinkB=5
            )
        )

    elif row["models"] == "MultiAdvEqOdds":
        ax.annotate(
            row['models'],
            xy=(row['micro_f1'], row['imparity_gender']),
            xytext=(32, 0),
            textcoords='offset points',
            ha='left',
            va='center',
            fontsize=8.5,
            arrowprops=dict(
                arrowstyle='-',
                color='0.4',
                lw=0.7,
                shrinkA=5,
                shrinkB=5
            )
        )

    else:
        ax.annotate(
            row['models'],
            xy=(row['micro_f1'], row['imparity_gender']),
            xytext=(8, 8),
            textcoords='offset points',
            ha='center',
            va='center',
            fontsize=8.5
        )

ax.set_xlabel(r'Micro-F1 $\uparrow$', fontsize=12)
ax.set_ylabel(r'Imparity $\downarrow$', fontsize=12)

ax.legend(
    title=None,
    frameon=True,
    loc='lower right'
)

sns.despine()

plt.tight_layout()
plt.show()
```
